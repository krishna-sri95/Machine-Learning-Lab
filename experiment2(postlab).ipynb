Perform preprocessing on Titanic Dataset

```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import LabelEncoder

# Load Titanic dataset
df = sns.load_dataset('titanic')

# Display first 5 rows
print("Original Dataset")
print(df.head())
```

```python
# Check missing values
print("\nMissing Values Before Preprocessing")
print(df.isnull().sum())

# Visualization of missing values
df.isnull().sum().plot(kind='bar')
plt.title(" [24EU02049] Missing Values Before Preprocessing")
plt.xlabel("Columns")
plt.ylabel("Count of Missing Values")
plt.show()
```

```python
# Fill missing values
df['age'] = df['age'].fillna(df['age'].median())
df['embarked'] = df['embarked'].fillna(df['embarked'].mode()[0])

# Drop columns with many missing values
df.drop(['deck', 'embark_town', 'alive', 'class', 'who'], axis=1, inplace=True)

# Check missing values after preprocessing
print("\nMissing Values After Preprocessing")
print(df.isnull().sum())
```

```python
# Encode categorical columns
le = LabelEncoder()

categorical_columns = ['sex', 'adult_male', 'alone', 'embarked']

for col in categorical_columns:
    df[col] = le.fit_transform(df[col].astype(str))

# Display preprocessed dataset
print("\nPreprocessed Dataset")
print(df.head())
```

```python
# Survival Distribution
df['survived'].value_counts().plot(kind='bar')
plt.title("[24EU02049] Survival Distribution")
plt.xlabel("Survived")
plt.ylabel("Count")
plt.show()

# Age Distribution
df['age'].hist()
plt.title("[24EU02049] Age Distribution After Preprocessing")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```
